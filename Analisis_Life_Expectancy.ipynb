```python
# Mencari nilai q1 dan q3
q1 = np.percentile(data['Adult Mortality'], 25)
q3 = np.percentile(data['Adult Mortality'], 75)

# Mencari nilai IQR
iqr = q3 - q1

# Mencari batas atas dan batas bawah
upper_bound_AM = q3 + 1.5 * iqr
lower_bound_AM = q1 - 1.5 * iqr

print("Upper Bound Adult Mortality:", upper_bound_AM)
print("Lower Bound Adult Mortality:", lower_bound_AM)
```

```python
outlier_AM = data['Adult Mortality'][:] > upper_bound_AM
outlier_AM2 = data['Adult Mortality'][outlier_AM].count()
print("Outlier dari Adult Mortality ada :", outlier_AM2)
```

Terdapat sebanyak 1 data outlier di variabel "Adult Mortality".

Dikarenakan distribusi dari "Adult Mortality" adalah distribusi miring kanan maka kami menggunakan imputasi median dikarenakan imputasi median  tahan terhadap pengaruh outlier karena nilai median tidak dipengaruhi oleh nilai ekstrem.

```python
# Mencari index data outlier
outliers = data[(data['Adult Mortality'] < lower_bound_AM) | (data['Adult Mortality'] > upper_bound_AM)].index

# Mengganti nilai outlier dengan NaN
data.loc[outliers, 'Adult Mortality'] = np.nan

# Mengisi nilai NaN dengan mean atau median
data['Adult Mortality'] = data['Adult Mortality'].fillna(data['Adult Mortality'] .median())

# Pengecekan
outlier_AM = data['Adult Mortality'][:] > upper_bound_AM
outlier_AM2 = data['Adult Mortality'][outlier_AM].count()
print("Outlier dari Adult Mortality ada :", outlier_AM2)
```

```python
x = data.drop(["Country","Life Expectancy"],axis=1)
y = data["Life Expectancy"]

from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(x)

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y,random_state = 0,test_size=0.25)

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn import linear_model

regr = linear_model.LinearRegression()
regr.fit(X_train,y_train)
y_pred = regr.predict(X_train)
print("R squared: {}".format(r2_score(y_true=y_train,y_pred=y_pred)))
```

#### Pengujian

Uji Barlett

H0 : Kedua kelompok data sampel yang telah diambil berasal dari populasi yang memiliki variansi yang sama.

H1 : Kedua kelompok data sampel yang telah diambil berasal dari populasi yang memiliki variansi yang berbeda.

Temuan: Karena p-value < 0.05, maka tolak H0. Artinya, "Schooling" dan "Adult Mortality" memiliki varians yang berbeda. Ini dapat diartikan sebagai adanya perbedaan dalam sebaran data antara kedua kelompok tersebut. Hasil tersebut dapat mengindikasikan heterokedastisitas.

Uji Goldfeld Quandt

H0 : Ragam bersifat homokedastis

H1 : Ragam bersifat heterokedastis

```python
import statsmodels.stats.api as sms
from statsmodels.compat import lzip
name = ['F statistic', 'p-value']
test = sms.het_goldfeldquandt(residuals, X_train)
lzip(name, test)
```

##### Pengujian

Uji Durbin Watson

H0: Tidak terdapat autokorelasi

H1: Terdapat autokorelasi

```python
import statsmodels.api as sm
dw = sm.stats.durbin_watson(residuals)
print('Durbin-Watson statistic:', dw)
```

Kriteria dalam pengujian Durbin Watson yaitu(Sujarweni, 2016: 232) :
1. Jika 0 < dw < dL, berarti ada autokorelasi positif
2. Jika 4 – dL < dw < 4, berarti ada auto korelasi negative
3. Jika 2 < dw < 4 – dU atau dU < dw < 2, berarti tidak ada autokorelasi positif atau
negatif
4. Jika dL ≤ dw ≤ dU atau 4 – dU ≤ dw ≤ 4 – dL, pengujian tidak meyakinkan. Untuk itu dapat digunakan uji lain atau menambah data
5. Jika nilai du < dw < 4-du maka tidak terjadi autokorelasi.

Didapatkan dL dan dU dari tabel dengan n = 183 dan k = 2 adalah dL = 1.7360 dan dU = 1.7802.

dw = 1.798281, kriteria yang sesuai dengan nilai nilai tersebut adalah kriteria kelima, artinya tidak terjadi autokorelasi maka asumsi tersebut terpenuhi.

##  **Asumsi Tidak Ada Multikolinearitas**

#### Heatmap

```python
korelasi = data.corr()
plt.figure(figsize=(7,7))
sns.heatmap(korelasi, annot=True, cmap='plasma')
plt.show()
```

Menurut Gujarati (2010):
1. Bila korelasi antara dua variabel bebas melebihi 0.8 maka multikolinearitas menjadi masalah yang serius.
2. Apabila korelasi antara variabel penjelas tidak lebih besar dibanding korelasi variabel terikat dengan masing-masing variabel penjelas, maka dapat dikatakan tidak terdapat masalah yang serius.

Nilai korelasi antara "Schooling" dan "Adult Mortality" adalah 0.52 yang mana Nilai tersebut < 0.8 dan kurang dari nilai korelasi antara variabel "Life Expectancy" dengan variabel-variabel independennya, Maka tidak terdapat masalah multikolinearitas.

Namun, ada cara mengecek multikolinearitas yang lebih powerfull yaitu menggunakan nilai VIF, sehingga pada langkah selanjutnya akan dicek asumsi multikolinearitas menggunakan nilai VIF.

#### Pengujian

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

X = data[['Schooling','Adult Mortality']]
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

vif = pd.DataFrame()
vif["Variable"] = X.columns
vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

print(vif)
```

Jika nilai VIF > 10 maka diasumsikan terjadi multikolinearitas antar variabel bebas. Terlihat pada output di atas bahwa nilai VIF dari kedua variabel bebas (Schooling dan Adult Mortality) yaitu sebesar 1.377184.
```python
# Impor pustaka yang diperlukan
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

X = data[['Schooling', 'Adult Mortality']]
y = data['Life Expectancy']

# Bagi data menjadi set pelatihan dan pengujian
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Pilih model regresi
model = LinearRegression()

# Latih model pada set pelatihan
model.fit(X_train, y_train)

# Lakukan prediksi pada set pengujian
y_pred = model.predict(X_test)

# Evaluasi model
mse = mean_squared_error(y_test, y_pred)
print(f'Mean Squared Error: {mse}')
```

# **Uji Simultan (Uji F) untuk Keseluruhan Model**

```python
import statsmodels.api as sm

# Menambahkan kolom konstan untuk intersep
data['intercept'] = 58.23

# Membuat model regresi berganda menggunakan statsmodels
X = data[['intercept', 'Schooling', 'Adult Mortality']]
y = data['Life Expectancy']
model = sm.OLS(y, X)
result = model.fit()

# Uji simultan (uji F) untuk keseluruhan model
f_statistic = result.f_test("Schooling = 0, Adult Mortality = 0")

# Mencetak hasil uji simultan
print(f'p-value: {f_statistic.pvalue}')

# Ambil keputusan berdasarkan nilai p
alpha = 0.05
if f_statistic.pvalue > alpha:
    print("Model tidak baik (terima H0)")
else:
    print("Model baik (tolak H0)")
```

# **Uji Parsial (Uji T)**

##  **Uji Parsial pada Koefisien β1 (Variabel Schooling)**

```python
t_statistic = result.t_test("Schooling = 0")

# Mencetak hasil uji parsial
print(f'p-value (Schooling): {t_statistic.pvalue}')

# Ambil keputusan berdasarkan nilai p
alpha = 0.05
if t_statistic.pvalue > alpha:
    print("Schooling tidak berpengaruh terhadap Life Expectancy (terima H0)")
else:
    print("Schooling berpengaruh terhadap Life Expectancy (tolak H0)")
```

##  **Uji Parsial pada Koefisien β2 (Variabel Adult Mortality)**

```python
t_statistic = result.t_test("Adult Mortality = 0")

# Mencetak hasil uji parsial
print(f'p-value (Adult Mortality): {t_statistic.pvalue}')

# Ambil keputusan berdasarkan nilai p
alpha = 0.05
if t_statistic.pvalue > alpha:
    print("Adult Mortality tidak berpengaruh terhadap Life Expectancy (terima H0)")
else:
    print("Adult Mortality berpengaruh terhadap Life Expectancy (tolak H0)")
```
